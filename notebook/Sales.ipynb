```python
import numpy as np

def is_str(val):
    if isinstance(val, str):
        val = val.replace('$', '').replace(',', '').strip()
        try:
            return float(val)
        except ValueError:
            return np.nan
    return val

R_sales['unit_price'] = R_sales['unit_price'].apply(is_str)
```
