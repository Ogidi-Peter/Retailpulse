```python
import numpy as np

def is_str(val):
    if isinstance(val, str):
        val = val.replace('₦', '').replace(',', '').strip()
        try:
            return float(val)
        except ValueError:
            return np.nan
    return val

R_sales['unit_price'] = R_sales['unit_price'].apply(is_str)
```

```python
R_sales['unit_price'] = R_sales['unit_price'].apply(is_str)
```

```python
parsed_slash = pd.to_datetime(R_sales['date'], format='%d/%m/%Y', errors='coerce')
parsed_iso = pd.to_datetime(R_sales['date'], format='%Y-%m-%d', errors='coerce')
```

```python
R_sales['date'] = parsed_slash.fillna(parsed_iso)
```

```python
R_sales=R_sales.dropna(subset=['unit_price'])
```

```python
R_sales['unit_price']
```

```python
R_sales = R_sales.drop_duplicates(keep='first')
```

```python
R_sales['returns'] =np.where(R_sales['quantity'] < 0, 'Return','Sales')
```

```python
R_sales.query('returns == "Return"').head(5)
```

```python
R_sales['payment_method'] = R_sales['payment_method'].str.lower().str.strip()
```

```python
R_sales['payment_method'].unique()
```

```python
R_sales['customer_age'] = R_sales['customer_age'].fillna(R_sales['customer_age'].median())
```

```python
def flag_outliers(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    upper_bound = q3 + 3 * iqr   # 3x IQR = strict outlier threshold
    return group > upper_bound

R_sales['price_outlier'] = R_sales.groupby('category')['unit_price'].transform(flag_outliers)
R_sales[R_sales['price_outlier']]   # inspect them
```

```python
R_sales.groupby('category')['unit_price'].describe()
```

```python
R_sales = R_sales[~R_sales['price_outlier']]
```

```python
R_sales = R_sales.drop(columns=['price_outlier'])
```

```python
R_sales.to_csv('../Data/processed/sales_clean.csv', index=False)
```
